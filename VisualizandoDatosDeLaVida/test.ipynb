```python
# LAs anotaciones de tipo son una forma de documentar el código y ayudar a los desarrolladores a entender
# qué tipos de datos se esperan como argumentos y qué tipo de dato se espera como resultado.
# Sin embargo, no son obligatorias y no afectan la ejecución del código.
# Ejemplo
def sum(x: int, y: int) -> bool:
    """
    Returns the sum of x and y.
    """
    if not isinstance(x, int) or not isinstance(y, int):
        raise TypeError("Both x and y must be integers.")
    return x + y

x: str  = 45
y: int = 2
result: int = sum(x, y)
print(f"The sum of {x} and {y} is {result}.")
```
