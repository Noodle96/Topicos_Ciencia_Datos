```python
from __future__ import annotations

from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt


PROJECT_ROOT: Path = Path("/home/russell/ssd/code/Topicos_Ciencia_Datos/Visual_Analytic_DEAP")

PROJECTIONS_DIR: Path = (
    PROJECT_ROOT
    / "dataset"
    / "processed"
    / "representations"
    / "manual_deap_features"
    / "projections"
)
```

```python
pca_df: pd.DataFrame = pd.read_csv(PROJECTIONS_DIR / "pca_2d.csv")
umap_df: pd.DataFrame = pd.read_csv(PROJECTIONS_DIR / "umap_2d.csv")
tsne_df: pd.DataFrame = pd.read_csv(PROJECTIONS_DIR / "tsne_2d.csv")

print("PCA:", pca_df.shape)
print("UMAP:", umap_df.shape)
print("t-SNE:", tsne_df.shape)

display(pca_df.head())
```

```python
def plot_projection(
    df: pd.DataFrame,
    title: str,
    color_column: str = "valence",
) -> None:
    plt.figure(figsize=(8, 6))

    scatter = plt.scatter(
        df["x"],
        df["y"],
        c=df[color_column],
        s=20,
        alpha=0.8,
    )

    plt.colorbar(scatter, label=color_column)
    plt.title(title)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.grid(True, alpha=0.3)
    plt.show()
```

```python
plot_projection(
    df=pca_df,
    title="PCA 2D - Manual DEAP Features",
    color_column="experiment_id",
)
```

```python
plot_projection(
    df=umap_df,
    title="UMAP 2D - Manual DEAP Features",
    color_column="experiment_id",
)
```

```python
plot_projection(
    df=tsne_df,
    title="t-SNE 2D - Manual DEAP Features",
    color_column="experiment_id",
)
```
