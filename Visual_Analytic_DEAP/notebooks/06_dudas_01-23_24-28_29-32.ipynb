```python
from __future__ import annotations

from pathlib import Path
from typing import Any

import mne
import numpy as np


BDF_DIR: Path = Path("../dataset/raw/bdf")


def inspect_status_candidates(participant_id: int) -> None:
    """
    Inspecciona los últimos canales de un participante DEAP
    y prueba cuáles contienen eventos tipo Status.
    """

    bdf_path: Path = BDF_DIR / f"s{participant_id:02d}.bdf"

    raw = mne.io.read_raw_bdf(
        bdf_path,
        preload=False,
        verbose=False,
    )

    print("=" * 80)
    print(f"Participant: s{participant_id:02d}")
    print(f"Num channels: {len(raw.ch_names)}")
    print("Last channels:")
    
    for index, channel_name in enumerate(raw.ch_names[-5:], start=len(raw.ch_names) - 5):
        print(f"{index + 1:02d}. '{channel_name}'")

    print("\nTesting last channels as possible Status channels:")

    for channel_name in raw.ch_names[-5:]:
        try:
            events: np.ndarray = mne.find_events(
                raw,
                stim_channel=channel_name,
                shortest_event=1,
                verbose=False,
            )

            event_codes, counts = np.unique(
                events[:, 2],
                return_counts=True,
            )

            print(f"\nChannel: '{channel_name}'")
            print(f"Events found: {len(events)}")
            print(dict(zip(event_codes.tolist(), counts.tolist())))

        except Exception as error:
            print(f"\nChannel: '{channel_name}'")
            print(f"Error: {error}")
```

```python
inspect_status_candidates(1)
inspect_status_candidates(24)
inspect_status_candidates(29)
inspect_status_candidates(32)
```
